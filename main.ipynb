```python
from groq import Groq
from dotenv import load_dotenv
import os

# .env file se variables load karo
load_dotenv()

# API key read karo
api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    raise ValueError("GROQ_API_KEY .env file mein nahi mili.")

# Groq client banao
client = Groq(api_key=api_key)

class PersonalAssistant:

    def __init__(self):
        print("HI, I am your AI assistant. How can I help")

    def ans_query(self):
        question = input("Ask me anythink:")

        response = client.chat.completions.create(
            model="qwen/qwen3.8-27b",
            messages=[
                {
                    "role":"system",
                    "content":"act like a helpful personal assistant."

                },
                {
                    "role":"user",
                    "content":question
                }
            ],
            temperature=0.7,
            max_tokens=512
        )
        print("\nAnswer:", response.choices[0].message.content.strip())

    def summarize_email(self):
        email_text = input("paste your email here:")

        response = client.chat.completions.create(
            model="qwen/qwen3.8-27b",
            messages=[
                {
                    "role":"system",
                    "content":(
                        "act like an expert email assistant."
                        "summarize the email in 2-3 sentences."
                    )
                },
                {
                    "role":"user",
                    "content":email_text
                }
            ],
            temperature=0.3,
            max_tokens=512
        )

        print(
            "\n Email Summary:",
            response.choices[0].message.content.strip()
        )

# create the assistant

assistant = PersonalAssistant()
```

```python
assistant.ans_query()
```

```python
assistant.summarize_email()
```
